### Task 1 (Batch Extraction & Boundary Inspection)
A sensor batch arrives as:

Extract the first reading, the final reading using negative indexing, and calculate the total count of readings using len().

[AUDIT] Batch Size: count | Baseline: first | Terminal: last

```python
# Task 1 solution:

raw_readings = [102.4, 105.1, 98.7, 110.2, 104.8, 99.1, 108.6, 103.5]

Baseline = raw_readings[0]  # Finds the first one in the list

Terminal = raw_readings[-1]  # Finds the last one in the list

count = len(raw_readings)   # Finds the count in the list


print(f"[AUDIT] Batch Size: {count} | Baseline: {Baseline} | Terminal: {Terminal} ")
```

### Task 6 (The Shallow Copy Mutation Trap)
Evaluate and predict what happens:


list_a = [10, 20, 30]

list_b = list_a

list_b.append(40)


What is list_a? Write the code, print list_a, and write a comment explaining the concept of shared memory references (pointer aliasing).

Then show how to create an independent copy so modifying list_b does not affect list_a.

```python
# Task 6 solution

list_a = [10, 20, 30]

list_b = list_a
# Direct assignment takes reference and not copy of data

list_b.append(40)

print(list_a)
# list_a got modified as the array is being shared to list_b so technically any append happened to list_b will happen to list_a

# To create an independent copy for making list_a unaffected.

list_b = list_a[:]
# Here we are taking copy of data

list_b.append(50)

print(list_a)

print(list_b)
```

### Task 8 (List Concatenation vs In-Place Extension)
Compare:


x = [1, 2]

x = x + [3, 4]  # Approach A

y = [1, 2]

y.extend([3, 4])  # Approach B

Both produce [1, 2, 3, 4]. Explain the difference in memory allocation and time complexity. (Hint: One instantiates a new list object; the other modifies the existing dynamic array buffer).

```python
# Task 8 Solution: Memory Allocation & Complexity Analysis
x = [1, 2]
x = x + [3, 4]  # Approach A: O(N + M) time & space. Creates a brand-new list object in the heap.

y = [1, 2]
y.extend([3, 4])  # Approach B: Amortized O(M) time, O(1) space. Modifies existing dynamic array buffer in-place.

print(f"Approach A: {x}")
print(f"Approach B: {y}")
```

### Task 9 (The Tuple Immutability Gotcha)
Run and analyze:

mixed_tuple = (1, 2, [30, 40])

mixed_tuple[2].append(50)

print(mixed_tuple)

Why does this execute successfully without raising a TypeError: 'tuple' object does not support item assignment? Explain what part of the tuple is immutable and what part is mutable.

```python
# Task 9 Solution: Tuple Mutability Mechanics
mixed_tuple = (1, 2, [30, 40])
mixed_tuple[2].append(50)

print(f"Mutated Tuple: {mixed_tuple}")

# Explanation:
# A tuple locks its internal memory pointers so addresses cannot be swapped.
# mixed_tuple[2] still points to the exact same list object in memory; since lists are mutable, modifying its contents succeeds.
```

### Task 10 (Reversing Sequences: [::-1] vs .reverse())
Given:

Pythonseries = [10, 20, 30, 40, 50]

Demonstrate two ways to reverse it:Using slicing [::-1].Using .reverse().Explain the difference in space complexity ($O(n)$ auxiliary memory vs $O(1)$ in-place).

```python
# Task 10 Solution: Slicing vs In-Place Reversal
series = [10, 20, 30, 40, 50]

# 1. Slicing: O(N) auxiliary space (creates a new reversed copy)
slicing_reversed = series[::-1]
print(f"Reversed via slice: {slicing_reversed}")

# 2. .reverse(): O(1) auxiliary space (mutates the list in-place and returns None)
series.reverse()
print(f"Reversed in-place: {series}")
```
