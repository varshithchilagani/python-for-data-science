### Task 4: Removing Duplicates with a Set
Input: A list of customer IDs who clicked an ad today:

clicked_ids = [101, 104, 101, 109, 112, 104, 115, 101]

Goal: Convert clicked_ids into a set named unique_clicks to remove duplicates.

Output: Print unique_clicks and print how many unique customers clicked using len().

```python
# Task 4 solution

clicked_ids = [101, 104, 101, 109, 112, 104, 115, 101]

unique_clicks = set(clicked_ids)

print(f"nunique_customers : {unique_clicks} \nunique_clicks : {len(unique_clicks)}")
```

### Task 5: Finding Common & Missing Customers (Set Math)

Input:

Python

jan_buyers = {101, 102, 103, 104}

feb_buyers = {103, 104, 105, 106}

Goal:

Find customers who bought in both months using .intersection().

Find customers who bought in January but NOT February (churned buyers) using .difference().

Output: Print both results with clear labels.

```python
# Task 5 solution

jan_buyers = {101, 102, 103, 104}

feb_buyers = {103, 104, 105, 106}

# Customers who bought in both the months
repeat_buyers = jan_buyers.intersection(feb_buyers)

print(f"Repeat Buyers: {repeat_buyers}")

# Customers who bought in January but not in February
churned_buyers = jan_buyers.difference(feb_buyers)

print(f"Churned Buyers: {churned_buyers}")
```

### Task 6: Why KeyError Happens
Goal: Create a small dictionary prices = {"apple": 50, "banana": 20}.

 Write a comment explaining what happens if you run prices["mango"] versus prices.get("mango", 0). Show the .get() version running cleanly.

```python
# Task 6 Solution: Safe Dictionary Lookups
prices = {"apple": 50, "banana": 20}

# Direct lookup prices["mango"] raises KeyError because the key does not exist.
# .get() safely returns the fallback default (0) without crashing the script.
mango_price = prices.get("mango", 0)
print(f"Mango Price: {mango_price}")
```

### Task 8: Valid vs. Invalid Dictionary Keys

Goal: Look at these three potential dictionary keys:

("Hyderabad", "Telangana") (a tuple)

["Hyderabad", "Telangana"] (a list)

404 (an integer)

Write a comment identifying which one will crash with TypeError: unhashable type and why (hint: mutability).

```python
# Task 8 Solution: Hashability & Dictionary Keys
# Valid keys (Immutable): ("Hyderabad", "Telangana") and 404
# Invalid key (Mutable): ["Hyderabad", "Telangana"] -> raises TypeError: unhashable type: 'list'
# Dictionary keys must be immutable so their hash value stays constant in memory.

valid_dict = {("Hyderabad", "Telangana"): "Capital Region", 404: "Not Found"}
print(valid_dict)
```

### Task 9: Speed Comparison (list vs set)

Goal: Explain in a comment why checking if 9999 in my_set: is much faster than checking if 9999 in my_list: when you have 1 million customer IDs. (Use the terms hash lookup / $O(1)$ vs linear scan / $O(n)$).

```python
# Task 9 Solution: Lookup Time Complexity
# Checking `9999 in my_list` performs an O(n) linear scan, checking items one by one from start to end.
# Checking `9999 in my_set` performs an O(1) hash lookup, computing hash(9999) to jump directly to its memory address in one step.
```

### Task 10: Counting Frequency (Classic Interview Pattern)
