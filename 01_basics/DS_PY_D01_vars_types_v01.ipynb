## Phase 1: Project Scenario — FinTech Merchant Settlement Pipeline

Context: You are working on the transaction settlement desk at a global fintech platform. Raw transactional events from partner point-of-sale (POS) systems arrive as uncleaned strings and mixed primitives. You must sanitize, calculate, and format the financial ledger entry.

### Task 1 (Raw Ingestion & Identification):
Receive three raw transaction values: merchant_name = "Urban_Retail_Hub", raw_tx_id = 908124, and is_settled = False. Store these, check and print the Python type of each variable using an f-string formatted as: "[AUDIT] Variable <var_name> is of type <type>".

```python
# Task 1 solution:
merchant_name = "Urban_Retail_Hub"
raw_tx_id = 908124  #given in the context
is_settled = False

print(f"{merchant_name} is of type {type(merchant_name)}")
print(f"{raw_tx_id} is of type {type(raw_tx_id)}")
print(f"{is_settled} is of type {type(is_settled)}")

```

### Task 2 (Data Cleaning & Float Casting):
The POS terminal sends transaction volume as a string with formatting: gross_amount_str = "12500.75". Convert this to a precise floating-point number in gross_amount. Then, create a constant PLATFORM_FEE_PCT = 0.025 (2.5%). Compute fee_deducted and store it rounded to 2 decimal places using round().

```python
# Task 2 solution:

gross_amount_str = "12500.75"
gross_amount = float(gross_amount_str)
PLATFORM_FEE_PCT = 0.025  # 2.5%

# final computation
fee_deducted = round(gross_amount * PLATFORM_FEE_PCT, 2)

print(fee_deducted)
```

### Task 3 (Ledger Balance Calculation):
Define net_settlement as gross_amount - fee_deducted. The merchant has a pre-existing reserve balance of 50000 (integer). Update total_merchant_reserve to add the net_settlement. Print the updated balance with commas as thousands separators (e.g., 62,188.23).

```python
# Task 3 solution:

net_settlement = gross_amount - fee_deducted
merchant_reserve_balance = 50000
total_merchant_reserve = merchant_reserve_balance + net_settlement


print(round(total_merchant_reserve),2)
```

### Task 4 (Flag Logic & String Masking):
Given a raw account identifier: account_num = "9876543210".
Create a masked string masked_account showing only the last 4 digits preceded by 6 asterisks: "******3210". Set a boolean kyc_cleared = True. Print an audit line: "Settlement approval for account ******3210: Status <kyc_cleared>".

```python
# Task 4 solution

account_num = "9876543210"
asterisks = "*" * 6
masked_account = asterisks + account_num[6:]
kyc_cleared = True
print(f"Settlement approval for account {masked_account}: {kyc_cleared}")
```

### Task 5 (Executive Summary Generation):
Compile an executive summary string using an f-string named settlement_summary that spans across multiple lines using triple quotes ("""..."""):
Include:

-> Merchant Name

-> Settlement Status

Display the output cleanly.

```python
# Task 5 solution:

Executive_summary = f"""
Merchant Name: {merchant_name}
Gross Amount: {gross_amount}
Net Settlement: {net_settlement}
Masked Account: {masked_account}
Settlement Status: {kyc_cleared}
"""

print(Executive_summary)
```
